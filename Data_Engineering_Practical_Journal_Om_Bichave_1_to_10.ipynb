### Task 1: Parse Text Files & Inspect Data Quality


```python
import pandas as pd

def parse_text_file(filepath):
    records = []
    with open(filepath, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith("#"):
                continue
            parts = line.split(",")
            record = {}
            for part in parts:
                if ":" in part:
                    k, v = part.split(":", 1)
                    record[k.strip()] = v.strip() if v.strip() != "None" else None
            if record:
                records.append(record)
    return pd.DataFrame(records)

# Example:
# df_txt = parse_text_file("sample.txt")
# inspect_data(df_txt, "TXT Document")
```

### Task 3: Search, Split & Replace Strings Using Regular Expressions


```python
import re

def search_pattern(text, pattern):
    return re.findall(pattern, text)

def split_pattern(text, pattern):
    return re.split(pattern, text)

def replace_pattern(text, pattern, replacement):
    return re.sub(pattern, replacement, text)

data = "Contact: om.bichave@email.com, Phone: 987-654-3210, Backup: admin@company.org"

emails = search_pattern(data, r"[\w.-]+@[\w.-]+\.\w+")
print("Emails found :", emails)

tokens = split_pattern(data, r"[,:\s]+")
print("Split tokens :", tokens)

masked = replace_pattern(data, r"\d{3}-\d{3}-\d{4}", "XXX-XXX-XXXX")
print("Masked text :", masked)
```

# Practical 4: Preprocessing, Data Normalization & Outlier Elimination

**Aim:** To write Python programs using Pandas, NumPy, Matplotlib, and Scikit-learn to handle missing values, duplicate records, data normalization, noise elimination, feature selection, and exploratory data analysis (EDA).

**Footer details:** Om Bichave | Roll No. 05 | Student ID 5175519 | T.Y. Data Science


### Part A – Missing Values, Duplicates & Min-Max Normalization


```python
import pandas as pd

data = {
    'Name': ['Amit', 'Riya', 'Amit', 'Neha', None, 'Riya'],
    'Age': [22, 24, 22, None, 26, 24],
    'Marks': [85, 90, 85, 78, None, 90]
}

df = pd.DataFrame(data)

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['Name'] = df['Name'].fillna('Unknown')
df = df.drop_duplicates()

df['Marks_Normalized'] = (df['Marks'] - df['Marks'].min()) / (df['Marks'].max() - df['Marks'].min())
df['Age_Normalized'] = (df['Age'] - df['Age'].min()) / (df['Age'].max() - df['Age'].min())

print(df[['Name', 'Age_Normalized', 'Marks_Normalized']])
```

### Part B – Outlier Elimination (IQR), Low Variance Selection & Feature Histograms


```python
from sklearn.feature_selection import VarianceThreshold

Q1, Q3 = df['Age'].quantile(0.25), df['Age'].quantile(0.75)
IQR = Q3 - Q1

df_clean = df[(df['Age'] >= Q1 - 1.5*IQR) & (df['Age'] <= Q3 + 1.5*IQR)]

selector = VarianceThreshold(threshold=0.0)
numeric_df = df_clean.select_dtypes(include='number')
selected_columns = numeric_df.columns[selector.fit(numeric_df).get_support()]

print("Selected Features:", selected_columns.tolist())
```

### Task 5: Implement Incremental Data Loading


```python
def load_incremental(new_data_df, db_conn, table_name):
    try:
        existing_ids = pd.read_sql(
            f"SELECT order_id FROM {table_name}", db_conn
        )["order_id"].tolist()
    except Exception:
        existing_ids = []

    incremental_df = new_data_df[
        ~new_data_df["order_id"].isin(existing_ids)
    ]

    if not incremental_df.empty:
        incremental_df.to_sql(
            table_name, db_conn, if_exists="append", index=False
        )
        print(f"Incremental update complete: {len(incremental_df)} new records added.")
    else:
        print("No new records to load.")

# Example:
# load_incremental(df_valid, conn, "validated_orders")
```

# Practical 10: Real-Time Data Streaming & Anomaly Detection Pipeline

**Aim:** To design and implement a streaming data pipeline with real-time aggregation, threshold-based alert systems, and automated anomaly detection for high-volume e-commerce transaction streams.

**Footer details:** Om Bichave | Roll No. 05 | Student ID 5175519 | T.Y. Data Science


### Task 1: Real-Time Data Stream Simulator & Sliding Window Aggregation


```python
print("Name: Om Bichave & Aryan Rane")

import pandas as pd
import numpy as np
import time
from datetime import datetime, timedelta

def generate_stream_batch(batch_id):
    np.random.seed(batch_id)
    n_records = 10

    timestamps = [
        datetime.now() - timedelta(seconds=i*2)
        for i in range(n_records)
    ]

    amounts = np.random.normal(
        loc=200, scale=50, size=n_records
    ).round(2)

    # Inject intentional anomaly
    if batch_id == 2:
        amounts[3] = 4500.00

    return pd.DataFrame({
        "timestamp": timestamps,
        "transaction_id": [
            f"TXN_{batch_id}_{i}" for i in range(n_records)
        ],
        "amount": amounts,
        "user_id": [f"U{100 + (i % 4)}" for i in range(n_records)]
    })

stream_buffer = pd.DataFrame()

for b_id in range(1, 4):
    batch = generate_stream_batch(b_id)
    stream_buffer = pd.concat(
        [stream_buffer, batch], ignore_index=True
    )

    window_summary = stream_buffer.groupby("user_id").agg(
        total_spent=("amount", "sum"),
        txn_count=("transaction_id", "count"),
        avg_spent=("amount", "mean")
    ).reset_index()

    print(f"--- Sliding Window Aggregation Batch {b_id} ---")
    print(window_summary.head())
```

### Task 2: Real-Time Anomaly Detection Engine (Z-Score & Threshold Monitoring)


```python
def detect_anomalies(df, threshold_z=2.5, max_amount_limit=3000):
    df = df.copy()
    df["mean"] = df["amount"].mean()
    df["std"] = df["amount"].std()
    df["z_score"] = (df["amount"] - df["mean"]) / df["std"]

    anomalies = df[
        (df["z_score"].abs() > threshold_z) |
        (df["amount"] > max_amount_limit)
    ].copy()

    anomalies["flag_reason"] = np.where(
        anomalies["amount"] > max_amount_limit,
        "Exceeded Static Monetary Limit",
        "Statistical Outlier (Z-Score)"
    )

    return anomalies[
        ["transaction_id", "user_id", "amount", "z_score", "flag_reason"]
    ]

detected_anomalies = detect_anomalies(stream_buffer)

print("Detected Real-Time Stream Anomalies:")
print(detected_anomalies)
```

### Task 3: Automated Alerting & Incident Log Persistence


```python
import sqlite3

conn = sqlite3.connect("stream_monitoring.db")

def route_alerts_and_persist(anomalies_df, db_conn):
    if not anomalies_df.empty:
        anomalies_df.to_sql(
            "incident_alerts", db_conn, if_exists="append", index=False
        )

        for _, row in anomalies_df.iterrows():
            print(
                f"[ALERT TRIGGERED] Txn {row['transaction_id']} "
                f"by User {row['user_id']} flagged! "
                f"Amount: ${row['amount']:.2f}. "
                f"Reason: {row['flag_reason']}"
            )
    else:
        print("[SYSTEM NORMAL] No anomalies detected in stream.")

route_alerts_and_persist(detected_anomalies, conn)
```
